# Visualization 1: Sleep Duration Distribution


```python
plt.figure()
plt.hist(S_H_L["Sleep Duration"], bins=15)
plt.xlabel("Sleep Duration (Hours)")
plt.ylabel("Number of Individuals")
plt.title("Distribution of Sleep Duration")
plt.show()
```

# Visualization 2: Stress Level vs Sleep Duration

```python
plt.figure()
plt.scatter(S_H_L["Stress Level"], S_H_L["Sleep Duration"],color='red')
plt.xlabel("Stress Level")
plt.ylabel("Sleep Duration (Hours)")
plt.title("Stress Level vs Sleep Duration")
plt.show()
```

 # Visualization 3: Physical Activity vs Sleep Quality


```python
plt.figure()
plt.scatter(S_H_L["Physical Activity Level"], S_H_L["Quality of Sleep"],color='orange')
plt.xlabel("Physical Activity Level")
plt.ylabel("Quality of Sleep")
plt.title("Physical Activity vs Sleep Quality")
plt.show()
```

```python
bmi_sleep = S_H_L.groupby("BMI Category")["Sleep Duration"].mean()
```

# Visualization 4: Average Sleep Duration by BMI Category

```python
bmi_sleep = S_H_L.groupby("BMI Category")["Sleep Duration"].mean()

plt.figure()
bmi_sleep.plot(kind="bar",color='Green')
plt.xlabel("BMI Category")
plt.ylabel("Average Sleep Duration (Hours)")
plt.title("Average Sleep Duration by BMI Category")
plt.show()
```

# Visualization 5: Daily Steps vs Heart Rate

```python
plt.figure()
plt.scatter(S_H_L["Daily Steps"], S_H_L["Heart Rate"],color='gray')
plt.xlabel("Daily Steps")
plt.ylabel("Heart Rate (bpm)")
plt.title("Daily Steps vs Heart Rate")
plt.show()
```

# Visualization 6: Sleep Disorder Distribution

```python
sleep_disorder_counts = S_H_L["Sleep Disorder"].value_counts()

plt.figure()
sleep_disorder_counts.plot(kind="pie", autopct="%1.1f%%")
plt.ylabel("")
plt.title("Distribution of Sleep Disorders")
plt.show()
```

# Additional Analysis: Occupation vs Sleep Duration

```python
occupation_sleep = S_H_L.groupby("Occupation")["Sleep Duration"].mean().sort_values()

plt.figure()
occupation_sleep.plot(kind="barh",color='purple')
plt.xlabel("Average Sleep Duration (Hours)")
plt.ylabel("Occupation")
plt.title("Average Sleep Duration by Occupation")
plt.show()
```
